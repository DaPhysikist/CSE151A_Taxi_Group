```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv('data/merged/train_processed.csv')
df.describe()
```

# Fare

```python
X = df.drop(columns=['trip_duration', 'fare_amount'])
y = df['fare_amount']

X_fare_train, X_fare_test, y_fare_train, y_fare_test = train_test_split(X, y, test_size=0.2, random_state=1)
scaler = StandardScaler()
X_fare_train_scaled = scaler.fit_transform(X_fare_train)
X_fare_test_scaled = scaler.transform(X_fare_test)
```

```python
fare_model = Sequential()

fare_model.add(Dense(64, input_dim=X_fare_train_scaled.shape[1], activation='relu'))
fare_model.add(Dense(32, activation='relu'))
fare_model.add(Dense(1, activation='linear'))

fare_model.compile(loss='mean_squared_error', optimizer=Adam(0.001))
fare_model.fit(X_fare_train_scaled, y_fare_train, epochs=10, batch_size=32, validation_split=0.2)
```

```python
y_fare_pred = fare_model.predict(X_fare_test_scaled)

print('Train MSE:', mean_squared_error(y_fare_train, fare_model.predict(X_fare_train_scaled)))
print('Test MSE:', mean_squared_error(y_fare_test, y_fare_pred))
print('R2:', r2_score(y_fare_test, y_fare_pred))
```

# Duration

```python
X = df.drop(columns=['trip_duration', 'fare_amount'])
y = df['trip_duration']

X_duration_train, X_duration_test, y_duration_train, y_duration_test = train_test_split(X, y, test_size=0.2, random_state=1)
scaler = StandardScaler()
X_duration_train_scaled = scaler.fit_transform(X_duration_train)
X_duration_test_scaled = scaler.transform(X_duration_test)
```

```python
duration_model = Sequential()

duration_model.add(Dense(64, input_dim=X_duration_train_scaled.shape[1], activation='relu'))
duration_model.add(Dense(32, activation='relu'))
duration_model.add(Dense(1, activation='linear'))

duration_model.compile(loss='mean_squared_error', optimizer=Adam(0.0001))
duration_model.fit(X_duration_train_scaled, y_fare_train, epochs=30, batch_size=32, validation_split=0.2)
```

```python
y_duration_pred = duration_model.predict(X_duration_test_scaled)

print('Train MSE:', mean_squared_error(y_duration_train, duration_model.predict(X_duration_train_scaled)))
print('Test MSE:', mean_squared_error(y_duration_test, y_duration_pred))
print('R2:', r2_score(y_duration_test, y_duration_pred))
```

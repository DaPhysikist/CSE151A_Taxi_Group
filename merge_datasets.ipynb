```python
durationDfSampled = pd.read_csv('data/duration/train_sampled.csv')
fareDfSampled = pd.read_csv('data/fare/train_sampled.csv')

N = 1000000
rounded_points = 3

durationDfSampled = durationDfSampled.sample(n=N, random_state=151)
fareDfSampled = fareDfSampled.sample(n=N, random_state=151)


print(durationDfSampled.columns, durationDfSampled.shape)
print(fareDfSampled.columns, fareDfSampled.shape)
```

```python
durationDfSampled['pickup_longitude_rounded'] = durationDfSampled['pickup_longitude'].round(rounded_points)
durationDfSampled['pickup_latitude_rounded'] = durationDfSampled['pickup_latitude'].round(rounded_points)
durationDfSampled['dropoff_longitude_rounded'] = durationDfSampled['dropoff_longitude'].round(rounded_points)
durationDfSampled['dropoff_latitude_rounded'] = durationDfSampled['dropoff_latitude'].round(rounded_points)

durationDfSampled.drop(['pickup_longitude', 'pickup_latitude', 'dropoff_longitude', 'dropoff_latitude', 'vendor_id'], axis=1, inplace=True)

durationDfSampled.head()
```

```python
fareDfSampled['pickup_longitude_rounded'] = fareDfSampled['pickup_longitude'].round(rounded_points)
fareDfSampled['pickup_latitude_rounded'] = fareDfSampled['pickup_latitude'].round(rounded_points)
fareDfSampled['dropoff_longitude_rounded'] = fareDfSampled['dropoff_longitude'].round(rounded_points)
fareDfSampled['dropoff_latitude_rounded'] = fareDfSampled['dropoff_latitude'].round(rounded_points)

fareDfSampled.drop(['pickup_longitude', 'pickup_latitude', 'dropoff_longitude', 'dropoff_latitude'], axis=1, inplace=True)

fareDfSampled.head()
```

```python
durationUnique = durationDfSampled['pickup_longitude_rounded'].nunique()
fareUnique = fareDfSampled['pickup_longitude_rounded'].nunique()
print(durationUnique, fareUnique)
```

```python
durationDfSampled = durationDfSampled.groupby(['pickup_longitude_rounded', 'pickup_latitude_rounded', 'dropoff_longitude_rounded', 'dropoff_latitude_rounded', 'passenger_count']).agg({'trip_duration': 'median'}).reset_index()
fareDfSampled = fareDfSampled.groupby(['pickup_longitude_rounded', 'pickup_latitude_rounded', 'dropoff_longitude_rounded', 'dropoff_latitude_rounded', 'passenger_count']).agg({'fare_amount': 'median'}).reset_index()

print(durationDfSampled.shape, fareDfSampled.shape)
```

```python
# Merge duration and fare df on rounded coordinates
mergedDf = pd.merge(durationDfSampled, fareDfSampled, on=['pickup_longitude_rounded', 'pickup_latitude_rounded', 'dropoff_longitude_rounded', 'dropoff_latitude_rounded', 'passenger_count'], how='inner')
mergedDf.describe()
```

```python
mergedDf.dropna(inplace=True)
mergedDf.drop_duplicates(inplace=True)
mergedDf.describe()
```

```python
mergedDf.to_csv('data/merged/train_processed.csv', index=False)
```
